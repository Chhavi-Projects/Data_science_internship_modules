```python
marks = [85, 90, 78, 92, 88]
student = {"name": "Chhavi", "age": 19, "marks": 85}

print("Marks:", marks)
print("Average:", sum(marks) / len(marks))
print("Student Details:", student)
print("Student Name:", student["name"])
```

```python
marks = [85, 90, 45, 72, 35]

for mark in marks:
    if mark >= 40:
        print(mark, "Pass")
    else:
        print(mark, "Fail")
```

```python
def calculate_average(numbers):
    return sum(numbers) / len(numbers)

marks = [85, 90, 78, 92, 88]
print("Average Marks:", calculate_average(marks))
```

```python
import pandas as pd

data = {"Name": ["Aman", "Riya", "Chhavi"],
        "Marks": [80, 90, 85]}

df = pd.DataFrame(data)

print(df)
print("Average Marks:", df["Marks"].mean())
print(df.describe())
```

```python
import pandas as pd

data = {"Product": ["Laptop", "Phone", "Tablet"],
        "Sales": [50000, 30000, 20000]}

df = pd.DataFrame(data)

print("Total Sales:", df["Sales"].sum())
print("Average Sales:", df["Sales"].mean())
print("Best-Selling Product:", df.loc[df["Sales"].idxmax(), "Product"])
```
