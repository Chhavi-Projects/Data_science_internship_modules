```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Dataset ready for visualization!")
```

```python
from google.colab import files
uploaded = files.upload()
df = pd.read_csv(next(iter(uploaded)))
```

```python
df.head()
```

Customer Segements Using Bar Charts.

```python
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x="customer_segment")
plt.title("Customer Distribution by Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Number of Customers")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

Acquistion Cost By Age Using Line Graph

```python
df["age_group"] = pd.cut(
    df["customer_age"],
    bins=[0, 25, 35, 45, 55, 65, 100],
    labels=["18-25", "26-35", "36-45", "46-55", "56-65", "66+"]
)

age_cost = df.groupby("age_group", observed=False)["customer_acquisition_cost"].mean()

plt.figure(figsize=(8, 5))
age_cost.plot(marker="o")
plt.title("Average Acquisition Cost by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Average Acquisition Cost")
plt.tight_layout()
plt.show()
```

Gender Distribution Using Pie Chart

```python
gender = df["gender"].value_counts()

plt.figure(figsize=(6, 6))
plt.pie(gender, labels=gender.index, autopct="%1.1f%%", startangle=90)
plt.title("Customer Distribution by Gender")
plt.show()
```

Customer Age Distribution Using Histograph

```python
plt.figure(figsize=(8, 5))
sns.histplot(data=df, x="customer_age", bins=15, kde=True)
plt.title("Distribution of Customer Ages")
plt.xlabel("Customer Age")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

Age VS Acquisition Cost Using Scatter Plot.

```python
plt.figure(figsize=(6, 4))
sns.scatterplot(data=df, x="customer_age", y="customer_acquisition_cost", alpha=0.5)
plt.title("Customer Age vs Acquisition Cost")
plt.xlabel("Customer Age")
plt.ylabel("Customer Acquisition Cost")
plt.tight_layout()
plt.show()
```

Print useful findings from your data.

```python
print("Largest Customer Segment:")
print(df["customer_segment"].value_counts().idxmax())

print("\nMost Common Age:")
print(df["customer_age"].mode().iloc[0])

print("\nAverage Acquisition Cost:")
print(round(df["customer_acquisition_cost"].mean(), 2))

print("\nCustomers by Gender:")
print(df["gender"].value_counts())

print("\nCorrelation: Age vs Acquisition Cost")
print(round(df["customer_age"].corr(df["customer_acquisition_cost"]), 3))
```
