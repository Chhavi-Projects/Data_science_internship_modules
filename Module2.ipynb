Import libraries and upload the CSV

```python
from google.colab import files

uploaded = files.upload()

```

```python
import pandas as pd
import numpy as np

filename = next(iter(uploaded))
df = pd.read_csv(filename)

print(df.head())
```

Check missing values and duplicates.

```python
print("Missing Values:\n", df.isnull().sum())
print("\nTotal Missing Values:", df.isnull().sum().sum())
print("\nDuplicate Rows:", df.duplicated().sum())
```

Explore the dataset.

```python
print("Dataset Shape:", df.shape)
print("\nColumn Names:", df.columns.tolist())
print("\nDataset Information:")
df.info()
print("\nFirst 5 Rows:")
display(df.head())
print("\nStatistical Summary:")
display(df.describe(include="all"))
```

Clean and organize the data.

```python
df.columns = df.columns.str.strip().str.lower().str.replace(" ", "_")
df = df.drop_duplicates()
df["customer_name"] = df["customer_name"].str.strip()
df["gender"] = df["gender"].str.strip()
df["customer_segment"] = df["customer_segment"].str.strip()

print("Data cleaned successfully!")
display(df.head())
```

Analyzing customer data using NumPy and Pandas.

```python
ages = df["customer_age"].to_numpy()
costs = df["customer_acquisition_cost"].to_numpy()

print("Average Customer Age:", np.mean(ages))
print("Minimum Age:", np.min(ages))
print("Maximum Age:", np.max(ages))
print("Average Acquisition Cost:", np.mean(costs))
print("Median Acquisition Cost:", np.median(costs))
print("Standard Deviation of Cost:", np.std(costs))
```

Analyzing customer segments.

```python
print("Customers by Segment:")
display(df["customer_segment"].value_counts())

print("\nAverage Acquisition Cost by Segment:")
display(df.groupby("customer_segment")["customer_acquisition_cost"].mean().round(2))
```

Analyzing Customer Geography.

```python
print("Customers by Country:")
display(df["customer_country"].value_counts())

print("\nCustomers by Region:")
display(df["region"].value_counts())

print("\nTop 10 Customer Cities:")
display(df["customer_city"].value_counts().head(10))
```

Analyzing Using Age groups.

```python
df["age_group"] = pd.cut(
    df["customer_age"],
    bins=[17, 25, 35, 45, 55, 65, 100],
    labels=["18-25", "26-35", "36-45", "46-55", "56-65", "66+"]
)

display(df["age_group"].value_counts().sort_index())
```
